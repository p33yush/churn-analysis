EDA Part 1: univariate analysis    

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv('../data/cleaned/customer_churn_cleaned.csv')
```

```python
churn_counts = df['churn'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(churn_counts, labels=['Retained (0)', 'Churned (1)'], autopct='%1.1f%%', colors=['#4CAF50', '#F44336'])
plt.title('Overall Churn Distribution')

plt.savefig('../images/churn_distribution.png', bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='contract', hue='churn', palette='Set2')
plt.title('Churn by Contract Type')
plt.savefig('../images/churn_by_contract.png')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='payment_method', hue='churn', palette='Set2')
plt.title('Churn by Payment Method')
plt.savefig('../images/churn_by_payment.png')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='tenure_group', hue='churn', palette='Set2')
plt.title('Churn by Tenure Group')
plt.savefig('../images/churn_by_tenure.png')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='monthlycharges', hue='churn', multiple='stack', bins=30)
plt.title('Distribution of Monthly Charges by Churn')
plt.savefig('../images/charges_dist.png')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='credit_score', hue='churn', multiple='stack', bins=30)
plt.title('Distribution of Credit Score by Churn')
plt.savefig('../images/credit_dist.png')
plt.show()
```
