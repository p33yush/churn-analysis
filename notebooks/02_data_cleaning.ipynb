```python
import pandas as pd
import numpy as np

df= pd.read_csv("../data/raw/customer_churn_1M.csv")
```

```python
df.shape
```

```python
df.head()
```

```python
df['signup_date'] = pd.to_datetime(df['signup_date'])
```

```python
df['signup_year']=df['signup_date'].dt.year
```

```python
df['signup_month']=df['signup_date'].dt.month
```

```python
df[['signup_date','signup_year','signup_month']].head()
```

```python

df['annual_income'] = df['annual_income'].fillna(df['annual_income'].median())
df['customer_satisfaction'] = df['customer_satisfaction'].fillna(df['customer_satisfaction'].median())
df['num_complaints'] = df['num_complaints'].fillna(df['num_complaints'].median())
df['avg_monthly_gb'] = df['avg_monthly_gb'].fillna(df['avg_monthly_gb'].median())
df['credit_score'] = df['credit_score'].fillna(df['credit_score'].median())

print("\nAfter:\n", df.isnull().sum())
```

```python
Q1 = df['annual_income'].quantile(0.25)
Q3 = df['annual_income'].quantile(0.75)
IQR = Q3-Q1
lower=Q1- 1.5*IQR
upper=Q3+ 1.5*IQR
print(lower)
print(upper)

((df['annual_income']<lower) | (df['annual_income']>upper)).sum()
```

```python
df['avg_monthly_spend'] = df['totalcharges'] / df['tenure']
print(df['avg_monthly_spend'].describe())
```

```python
df['tenure_group'] = pd.cut(
    df['tenure'], 
    bins=[0, 12, 24, 48, 72], 
    labels=['New', 'Mid', 'Established', 'Loyal']
)
print(df['tenure_group'].value_counts())
```

```python
median_charges = df['monthlycharges'].median()
df['is_high_value'] = (df['monthlycharges'] > median_charges).astype(int)
print(df['is_high_value'].value_counts())
```

```python
df['total_support_interactions'] = df['num_complaints'] + df['num_service_calls']
print(df['total_support_interactions'].describe())
```
