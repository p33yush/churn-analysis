```python
import duckdb
import pandas as pd

# Connect to our new DuckDB file
conn = duckdb.connect('../data/churn_database.duckdb')
```

```python
query = """
SELECT 
    COUNT(*) as total_customers,
    AVG(churn) * 100 as churn_rate_percent
FROM customers;
"""
df_result = conn.execute(query).df()
print(df_result)
```
